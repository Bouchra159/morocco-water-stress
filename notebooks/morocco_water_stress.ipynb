# Running Dry: Mapping Water Stress in Morocco

A short, reproducible walkthrough of Morocco's water stress using open data.
We look at the national trend since 1961, then zoom into the **Oum Er-Rbia basin**
whose Al Massira reservoir supplies Casablanca and the Doukkala irrigated plain.

All data is public and openly licensed. Run `src/fetch_worldbank.py` and
`src/fetch_geodata.py` first to populate `data/`.

```python
import pandas as pd

df = pd.read_csv('../data/processed/morocco_water_indicators.csv', index_col='year')
df.tail()
```

## 1. Morocco crossed the water-stress line around 2000

Renewable internal freshwater per person fell from **2,431 m³ (1961)** to
**777 m³ (2022)** — below the international water-stress threshold of 1,000 m³,
and heading toward the 500 m³ 'absolute scarcity' line.

```python
pc = df['renewable_internal_pc_m3'].dropna()
print(f'{pc.iloc[0]:.0f} m3 ({pc.index[0]})  ->  {pc.iloc[-1]:.0f} m3 ({pc.index[-1]})')
print(f'decline: {100*(1 - pc.iloc[-1]/pc.iloc[0]):.0f}%')
```

![per-capita decline](../figures/fig1_percapita_decline.png)

## 2. Same water, three times the people

Total renewable internal freshwater has stayed flat (~29 billion m³/year).
The per-capita fall is driven by population growth, not a shrinking resource —
with drought making the *usable* share smaller still.

```python
total = df['renewable_internal_total_bcm'].dropna()
pop = df['population'].dropna() / 1e6
print(f'renewable water: {total.iloc[0]:.0f} -> {total.iloc[-1]:.0f} billion m3 (flat)')
print(f'population: {pop.iloc[0]:.1f}M -> {pop.iloc[-1]:.1f}M')
```

![resource vs people](../figures/fig2_resource_vs_people.png)

## 3. Where the water goes: agriculture

Agriculture accounts for about **88%** of all freshwater withdrawn (2022).
That makes water stress inseparable from farming policy and rural livelihoods.

```python
ag = df['withdrawal_agriculture_pct'].dropna()
print(f'agriculture share of withdrawals ({ag.index[-1]}): {ag.iloc[-1]:.0f}%')
```

![agriculture share](../figures/fig3_agriculture_share.png)

## 4. The drought is visible in the dams

Reported national reservoir fill rates fell to ~23% in 2024. These figures are
reported by Moroccan authorities/press (not the World Bank) and are kept in a
clearly-labelled separate file.

```python
res = pd.read_csv('../data/raw/reservoir_levels_reported.csv', comment='#')
res[['date', 'national_fill_pct']]
```

![reservoir levels](../figures/fig4_reservoir_levels.png)

## 5. Measuring it from space: the Al Massira reservoir

Instead of citing a reported fill rate, we **measure** the reservoir's
open-water surface directly from Sentinel-2 imagery — NDWI water index with
an automatic Otsu threshold, dry-season scenes, cloud-masked. Al Massira
supplies Casablanca and the Doukkala irrigated plain.

This uses public Sentinel-2 L2A via the Earth Search STAC API (no Earth
Engine account). See `src/measure_reservoir.py`.

```python
res_area = pd.read_csv('../data/processed/reservoir_area.csv')
peak = res_area['water_area_km2'].max()
trough = res_area['water_area_km2'].min()
print(f'water surface: {peak:.0f} km2 (peak) -> {trough:.0f} km2 (trough)')
print(f'loss: {100*(1-trough/peak):.0f}%')
res_area
```

Middle Atlas, is stored behind the Al Massira Dam, and supplies Casablanca and
the Doukkala plain. The terrain map is scripted with PyQGIS over OpenTopoMap.
